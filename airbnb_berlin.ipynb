```python
#checking for duplicates

df.isna().sum()
df.duplicated().sum()
df['id'].duplicated().sum()
```

```python
df['last_review'] = pd.to_datetime(df['last_review'])
```

```python
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
df['host_name'] = df['host_name'].fillna('Unknown')
df['license'] = df['license'].fillna('No license info')
```

```python
# Converting price to a number; anything unparseable becomes NaN (missing)
df['price'] = pd.to_numeric(df['price'], errors='coerce')

# Converting last_review text into a real date; blanks/bad values become NaT (missing date)
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')
```

```python
df.info()
```

```python
#cleaning the text columns

text_cols = ['name', 'host_name', 'neighbourhood_group', 'neighbourhood', 'room_type']

for col in text_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
# 1. The summary — watch the gap between 75% and max
print(df['price'].describe())

# 2. The top values — are the extremes real or junk?
print(df['price'].sort_values(ascending=False).head(15).tolist())

# 3. The percentiles — where does the data stop being normal?
print(df['price'].quantile([0.95, 0.99, 0.995, 0.999]))
```

```python
df['price'].describe()
df[df['price'] > 1000]
```

```python
#remove prices that seem unrealistic such as 50k, 40k

df.loc[df['price'] > 5000, 'price'] = pd.NA
```

```python
#Creating columns
```

```python
def price_category(price):
    if pd.isna(price):
        return 'Missing'
    elif price < 75:
        return 'Budget'
    elif price < 150:
        return 'Mid-range'
    elif price < 300:
        return 'Premium'
    else:
        return 'Luxury'

df['price_category'] = df['price'].apply(price_category)
```

```python
def availability_category(days):
    if days == 0:
        return 'Unavailable'
    elif days <= 90:
        return 'Low availability'
    elif days <= 180:
        return 'Medium availability'
    else:
        return 'High availability'

df['availability_category'] = df['availability_365'].apply(availability_category)
```

```python
df['has_license'] = df['license'].ne('No license info')
```

```python
df['host_type'] = df['calculated_host_listings_count'].apply(
    lambda x: 'Single listing host' if x == 1 else 'Multi-listing host'
)
```

```python
df['has_reviews'] = df['number_of_reviews'] > 0
```
